```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
from matplotlib.pyplot import rc
import matplotlib.font_manager

rc('font',**{'size':'22','family':'serif','serif':['CMU serif']})
rc('mathtext', **{'fontset':'cm'})
rc('text', usetex=True)
rc('legend',**{'fontsize':'18'})
matplotlib.rcParams['axes.linewidth'] = 1.5
matplotlib.rcParams['axes.labelsize'] = 20
matplotlib.rcParams['xtick.labelsize'] = 15
matplotlib.rcParams['ytick.labelsize'] = 15
matplotlib.rcParams['legend.fontsize'] = 15

matplotlib.rcParams['axes.titlesize'] = 30
matplotlib.rcParams['text.latex.preamble'] = r'\usepackage{amsmath}'

plt.rc('xtick', direction='in')
plt.rc('ytick', direction='in')

# Additional options to have minor ticks
from matplotlib.ticker import AutoMinorLocator
```

```python
save_plot = 1

GaussCov = 0

len_fnl = 10
fnl_lo_arr = np.logspace(0, 5, len_fnl, base=10)

if GaussCov == 0 :
    gausscov_tag = ''
elif GaussCov == 1:
    gausscov_tag = '_%s' %'GaussCov'
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 5))  # 2 rows, 2 columns
fig.subplots_adjust(hspace=0.02, wspace=0.3)  # Adjust vertical and horizontal space

##############################
###### NGC

sky = 'N'

if sky == 'N':
    lenP0 = 45
    lenP0B0 = lenP0 + 562
    sky_str = ''
elif sky == 'S':
    lenP0 = 55
    lenP0B0 = lenP0 + 564
    sky_str = '_SGC'

# load theory
SNR_cos_s1 = np.loadtxt('SNR_PB_cos_s1%s%s.dat' %(gausscov_tag, sky_str))
SNR_cos_s2 = np.loadtxt('SNR_PB_cos_s2%s%s.dat' %(gausscov_tag, sky_str))
SNR_cos_s3 = np.loadtxt('SNR_PB_cos_s3%s%s.dat' %(gausscov_tag, sky_str))
SNR_cos_rot_cPB_s1 = np.loadtxt('SNR_PB_cos_cPB_s1%s%s.dat' %(gausscov_tag, sky_str))
SNR_cos_rot_s1 = np.loadtxt('SNR_PB_cos_PcB_s1%s%s.dat' %(gausscov_tag, sky_str))


# plot
ax[0].plot(fnl_lo_arr, SNR_cos_s1[:, -1], label = '$k_f$')
ax[0].plot(fnl_lo_arr, SNR_cos_s2[:, -1], linestyle = 'dotted', label = '$2~k_f$')
ax[0].plot(fnl_lo_arr, SNR_cos_s3[:, -1], linestyle = 'dashed', lw = 2, label = '$3~k_f$')
ax[0].plot(fnl_lo_arr, SNR_cos_rot_s1[:, lenP0+9], linestyle = 'dashdot', label = '$\mathrm{PcB}~9$')

#ax1.legend(fontsize = 12, loc = (1.02, 0.5), ncol = 1)
ax[0].set_xlabel('$f_\mathrm{NL}$')
ax[0].set_ylabel(r'$\cos\Big(\mathbf{T}(f_\mathrm{NL}), \mathbf{D}\Big)$')
ax[1].set_ylabel(r'$\cos\Big(\mathbf{T}(f_\mathrm{NL}), \mathbf{D}\Big)$')
ax[0].set_xscale('log')
ax[0].set_ylim(0.4, 0.97)


ax[0].set_title("$\mathrm{NGC}$", fontsize=20)

##############################
###### SGC

sky = 'S'

if sky == 'N':
    lenP0 = 45
    lenP0B0 = lenP0 + 562
    sky_str = ''
elif sky == 'S':
    lenP0 = 55
    lenP0B0 = lenP0 + 564
    sky_str = '_SGC'

# load theory
SNR_cos_s1 = np.loadtxt('SNR_PB_cos_s1%s%s.dat' %(gausscov_tag, sky_str))
SNR_cos_s2 = np.loadtxt('SNR_PB_cos_s2%s%s.dat' %(gausscov_tag, sky_str))
SNR_cos_s3 = np.loadtxt('SNR_PB_cos_s3%s%s.dat' %(gausscov_tag, sky_str))
SNR_cos_rot_cPB_s1 = np.loadtxt('SNR_PB_cos_cPB_s1%s%s.dat' %(gausscov_tag, sky_str))
SNR_cos_rot_s1 = np.loadtxt('SNR_PB_cos_PcB_s1%s%s.dat' %(gausscov_tag, sky_str))


# plot
ax[1].plot(fnl_lo_arr, SNR_cos_s1[:, -1], label = '$P + B, \Delta k$')
ax[1].plot(fnl_lo_arr, SNR_cos_s2[:, -1], linestyle = 'dotted', label = '$P + B, 2\,\Delta k$')
ax[1].plot(fnl_lo_arr, SNR_cos_s3[:, -1], linestyle = 'dashed', lw = 2, label = '$P + B, 3\,\Delta k$')
ax[1].plot(fnl_lo_arr, SNR_cos_rot_s1[:, lenP0+9], linestyle = 'dashdot', label = '$P + cB $')

ax[1].set_xlabel('$f_\mathrm{NL}$')
ax[1].set_xscale('log')
ax[1].set_ylim(0.4, 0.97)

ax[1].set_title("$\mathrm{SGC}$", fontsize=20)
ax[1].legend(prop={'size':20}, ncol=1, loc = (1.04, 0.5))

#plt.savefig('cosine_plot.pdf', dpi = 500, bbox_inches = 'tight')
```
