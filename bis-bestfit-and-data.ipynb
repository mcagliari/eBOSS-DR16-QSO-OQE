```python
def plot_best_fit(Bkd, cov, Bkf, save=None, color="blue"):
    plt.errorbar(np.arange(len(Bkd)), np.abs(Bkd), yerr=np.sqrt(cov), ls='none', marker='.', label="data", color=color)
    plt.semilogy(np.abs(Bkf), linestyle = 'dashdot', label="bestfit", color=color)
    
    plt.legend(frameon=False)
    plt.xlabel('Triangle index')
    plt.ylabel('$|B_0(k)|$')
    plt.tight_layout()
    
    if save is not None:
        plt.savefig(save, bbox_inches='tight')
    plt.show()
    
def plot_comparison(Bkdfkp, covfkp, Bkffkp, kd, Bkd, cov, kf, Bkf, save=None, colorfkp="orange", color="blue"):
    
    plt.errorbar(np.arange(len(Bkdfkp)), np.abs(Bkdfkp), yerr=np.sqrt(covfkp), ls='none', marker='s', markersize=3, label="FKP", color=colorfkp)
    plt.semilogy(np.abs(Bkffkp), linestyle = 'dashed', label="FKP", color=colorfkp)
    
    plt.errorbar(np.arange(len(Bks)), np.abs(Bkd), yerr=np.sqrt(cov), ls='none', marker='.', label="Optimal", color=color)
    plt.semilogy(np.abs(Bkf), linestyle = 'dashdot', label="Optimal", color=color)
    
    plt.legend(frameon=False)
    plt.xlabel('Triangle index')
    plt.ylabel('$|B_0(k)|$')
    plt.tight_layout()
    
    if save is not None:
        plt.savefig(save, bbox_inches='tight')
    plt.show()
```

# Table plot

```python
def plot_ax(ax, Bkd, Bkf, cov, linestyle, marker, color, annotate=None, xy=(20,20)):
    ax.errorbar(np.arange(len(Bkd)), np.abs(Bkd), yerr=np.sqrt(cov), 
                 ls='none', marker=marker, markersize=4, label='Data', color=color)
    ax.semilogy(np.abs(Bkf), 
               linestyle = linestyle, color=color, label='Best fit')
    
    ax.legend(loc=1,frameon=False)
    ax.set_xlabel('Triangle index')
    ax.set_ylabel('$|B_0(k)|$')
    
    if annotate is not None:
        ax.annotate(annotate, xy=xy, xycoords='axes pixels')
```

```python
fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(6*2, 4*1))

cN = 'darkblue'
cS = 'tomato'


Bkd = np.loadtxt("bis-fits/input-s3rebin/data/bispectra/NGC/Bk_data.dat")
cov = np.diagonal(np.loadtxt("bis-fits/input-s3rebin/data/bispectra/NGC/covariance_SNsub_NGC_B.dat"))

Bkf = np.loadtxt("bestfit_Bk/Bk_jointN_Opt_p1p0.dat").T

plot_ax(ax[0], Bkd[:,4], Bkf[:,3], cov, 'dashed', 's', cN, None, xy=(270,170))
ax[0].set_title('NGC')

##################

Bkd = np.loadtxt("bis-fits/input-s3rebin/data/bispectra/SGC/Bk_data.dat")
cov = np.diagonal(np.loadtxt("bis-fits/input-s3rebin/data/bispectra/SGC/covariance_SNsub_SGC_B.dat"))

Bkf = np.loadtxt("bestfit_Bk/Bk_jointS_Opt_p1p0.dat").T

plot_ax(ax[1], Bkd[:,4], Bkf[:,3], cov, 'dashed', 's', cS, None, xy=(270,170))
ax[1].set_title('SGC')

plt.savefig("bis-figures/Bk_bestfit_joint_Opt_p1p0.pdf", bbox_inches='tight')
plt.show()
```
